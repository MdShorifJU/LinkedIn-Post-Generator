```python
import os
from getpass import getpass

from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

print("Libraries imported successfully!")
```

```python


os.environ["GROQ_API_KEY"] = getpass("Enter Groq API Key: ")
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0.7
)
```

```python
tech_writer_prompt = ChatPromptTemplate.from_template("""
You are a professional LinkedIn technology content writer.

Your task is to write an engaging LinkedIn post about the given topic.

Requirements:
- Write 2-4 short paragraphs.
- Use a professional and engaging tone.
- Explain the topic clearly.
- Include relevant insights.
- Use suitable emojis where appropriate.
- End with a thoughtful question or call-to-action.
- Write entirely in the requested language.

Topic: {topic}
Language: {language}

Generate only the LinkedIn post.
""")

tech_writer_chain = tech_writer_prompt | llm | StrOutputParser()
```

```python
general_writer_prompt = ChatPromptTemplate.from_template("""
You are a professional LinkedIn content writer.

Your task is to write an engaging LinkedIn post about
a general topic such as education, business, career,
leadership, productivity, or personal development.

Requirements:
- Write 2-4 short paragraphs.
- Use a professional and engaging tone.
- Provide meaningful insights.
- Use suitable emojis where appropriate.
- End with a thoughtful question or call-to-action.
- Write entirely in the requested language.

Topic: {topic}
Language: {language}

Generate only the LinkedIn post.
""")

general_writer_chain = general_writer_prompt | llm | StrOutputParser()
```

```python
tech_post = tech_writer_chain.invoke({
    "topic": "Artificial Intelligence in Healthcare",
    "language": "English"
})

print("TECH LINKEDIN POST")
print("=" * 50)
print(tech_post)
```

```python
general_post = general_writer_chain.invoke({
    "topic": "Importance of Education in Building a Better Future",
    "language": "Bengali"
})

print("GENERAL LINKEDIN POST")
print("=" * 50)
print(general_post)
```

```python
router_prompt = ChatPromptTemplate.from_template("""
You are a topic classification agent.

Classify the topic into exactly one category:

Tech:
AI, Machine Learning, Programming, Software,
Cybersecurity, Robotics, Data Science, Technology.

General:
Education, Business, Career, Leadership,
Productivity, Lifestyle, Personal Development.

Topic: {topic}

Return ONLY one word:
Tech or General
""")

router_chain = router_prompt | llm | StrOutputParser()
```

```python
def linkedin_post_generator(topic, language):

    # Step 1: Classify topic
    category = router_chain.invoke({
        "topic": topic
    }).strip()

    print("Topic:", topic)
    print("Category:", category)

    # Step 2: Conditional handover
    if category.lower() == "tech":
        print("Selected Agent: Tech Writer Agent")

        result = tech_writer_chain.invoke({
            "topic": topic,
            "language": language
        })

    else:
        print("Selected Agent: General Writer Agent")

        result = general_writer_chain.invoke({
            "topic": topic,
            "language": language
        })

    return result
```

```python
post = linkedin_post_generator(
    topic="Artificial Intelligence in Healthcare",
    language="English"
)

print("\nGenerated LinkedIn Post:")
print("=" * 50)
print(post)
```

```python
post = linkedin_post_generator(
    topic="Importance of Education in Building a Better Future",
    language="Bengali"
)

print("\nGenerated LinkedIn Post:")
print("=" * 50)
print(post)
```
